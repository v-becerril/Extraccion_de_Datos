```python
#Importamos librerías requeridas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Convertir en dataframe el archivo CSV limpio con las variables de interés (50 columnas)
df = pd.read_csv("Bergamo_Italia.csv")
df.head(4)
```

```python
#Carga de CSV original 
df_original= pd.read_csv("listings.csv")
df_original.head(4)
```

APLICACIÓN DE FILTROS

```python
#FiltroA = Resgisto de 5 "host_name" diferentes 
#En este caso se utilizó la base de datos original, ya que en la base de datos tratada decidí eliminar por temas más prácticos esta columna, para los siguientes filtros no fue necesario utilizar la base de datos original. 
#Se utilizó la función "value_counts()" para contar la cantidad de veces que se repite cada "host_name" y luego se utilizó "head(5)" para mostrar los 5 más frecuentes.
host_name = df_original['host_name'].value_counts().head(5)
host_name
```

```python
#Exportar filtro "host_name" a CSV
host_name.to_csv("FiltroA.csv")
```

```python
#FiltroB = Resgistro de los host que se hallan unido a Airbnb después del año 2020
#Se utilizó el filtro de comporación para seleccionar las fechas posteriores al 31 de diciembre de 2020.
incorporacion = df[df['host_since'] > '2020-12-31']
incorporacion
```

```python
#Corroboración que el filtro se haya aplicado correctamente
#Se utilizó la función "min()" para verificar que la fecha mínima en la columna "host_since" sea posterior al 31 de diciembre de 2020.
incorporacion['host_since'].min()
```

```python
#Exportar filtro incorporacion a CSV
incorporacion.to_csv("FiltroB.csv")
```

```python
#Verificación de los tiempos de respuesta de los host
#Se utilizó la función "value_counts()" para contar la cantidad de veces que se repite cada tiempo de respuesta.
host_response_time = df['host_response_time'].value_counts()
host_response_time
```

```python
#FiltroC = Registro de los host que responden a más tardar en 1 día
#Se utilizó el filtro lógico con contenedores para filtrar los tiempos de respuesta "within an hour", "within a few hours" y "within a day", unidos por el operador OR "|".
respuesta = df[(df['host_response_time'] == 'within an hour') | (df['host_response_time'] == 'within a day') | (df['host_response_time'] == 'within a few hours')]
respuesta
```

```python
#Verificación de los tiempos de respuesta de los host en el FiltroC
c = respuesta['host_response_time'].value_counts()
c
```

```python
#Exportar filtro respuesta a CSV
respuesta.to_csv("FiltroC.csv")
```

```python
#FiltroD = Los registros de las filas “Múltiplos de 200… (400, 600, 800, etc.)”
#Se utilizó la función "iloc" para seleccionar las filas que son múltiplos de 200, utilizando el fomato de columnas consecutivas, es decir, comenzando desde la fila 0 y seleccionando cada 200 filas.
multiplos = df.iloc[::200, :]
multiplos
```

```python
#Exportar filtro multiplos a CSV
multiplos.to_csv("FiltroD.csv")
```

```python
#Verificación de los tipos de cuartos
#Se utilizó la función "value_counts()" para contar la cantidad de veces que se repite cada tipo de cuarto.
room_type = df['room_type'].value_counts()
room_type
```

```python
#FiltroE = Los registros de los tipos de cuarto: “Entire home”  y “Private room”
#Se utilizó el filtro lógico con contenedores para filtrar los tipos de cuarto "Entire home/apt" y "Private room", unidos por el operador OR "|".
cuarto = df[(df['room_type'] == 'Entire home/apt') | (df['room_type'] == 'Private room')]
cuarto
```

```python
#Verificación de los tipos de cuartos en el FiltroE
d = cuarto['room_type'].value_counts()
d
```

```python
#Exportar filtro cuarto a CSV
cuarto.to_csv("FiltroE.csv")
```

```python
#Verificación el número de camas
#Se utilizó la función "value_counts()" para contar la cantidad de veces que se repite un número de camas.
beds = df['beds'].value_counts()
beds
```

```python
#Filtro F = Los registros que “a lo mucho cuenten con 4 camas”
#Se utilizó el filtro de comparación para filtrar los registros que tienen 4 camas o menos.
camas = df[df['beds'] <= 4]
camas
```

```python
#Verificación el número de camas en el FiltroF
f = camas['beds'].value_counts()
f
```

```python
#Exportar filtro camas a CSV
camas.to_csv("FiltroF.csv")
```

```python
#Verificación de los tipos de host
host = df['host_is_superhost'].value_counts()
host
```

```python
#FiltroG = Los registros que “sean superhost”
#Se utilizó el filtro lógico para filtrar los registros que tienen el valor "t" en la columna "host_is_superhost".
superhost = df[df['host_is_superhost'] == 't']
superhost
```

```python
#Verificación de los tipos de host
g = superhost['host_is_superhost'].value_counts()
g
```

```python
#Exportar filtro superhost a CSV
superhost.to_csv("FiltroG.csv")
```

```python
#FiltroH = Los registros de las columnas “impares”
#Se utilizó la función "iloc" para seleccionar las columnas impares, comenzando desde la columna 1 (segunda columna) y seleccionando cada 2 columnas, se utilizó ":" para seleccionar todas las filas y "1::2" para seleccionar las columnas impares.
#Se inicio desde la segunda columna, ya que la primera columna es la 0 y "sources" es la primera columna seleccionada del dataframe original.
impares = df.iloc[:, 1::2]
impares
```

```python
#Exportar filtro impares a CSV
impares.to_csv("FiltroH.csv")
```
